Calcule usando Numpy:

1. **Análise Estatística:** Calcule a média, mediana e desvio padrão das temperaturas usando Numpy.

```python
def calculando(temperaturas):

    media = np.mean(temperaturas).round(3)
    mediana = np.median(temperaturas).round(3)
    desvio = np.std(temperaturas).round(3)
    return media, mediana, desvio

rtmedia,rtmediana,rtdesvio=calculando(temperaturas)

print(f"Média: {rtmedia} | Mediana: {rtmediana} | Desvio padrão: {rtdesvio}")

```

2. **Classificação de Temperaturas:** Classifique cada dia como "frio", "moderado" ou "quente". Você pode definir os limites para cada categoria. Por exemplo:
    - Frio: Temperatura abaixo de 5 graus.
    - Moderado: Temperatura entre 5 e 15 graus.
    - Quente: Temperatura acima de 15 graus.

```python
for temperatura in temperaturas:
    if temperatura <= 10:
        print(f"{temperatura}: Frio")
    elif temperatura > 10 and temperatura <= 25:
        print(f"{temperatura}: Moderado")
    else:
        print(f"{temperatura}: Quente")
```

Dias Extremos: Identifique os dias mais frios e mais quentes do ano.

```python
print(f"Tem {temperaturas.shape[0]} dias ao total")
diaFrio = np.argmin(temperaturas)
diaQuente = np.argmax(temperaturas)


temp_minima = temperaturas[diaFrio]
temp_maxima = temperaturas[diaQuente]


print(f"Dia mais frio: dia {diaFrio} (Temperatura: {temp_minima}°C)")
print(f"Dia mais quente: dia {diaQuente} (Temperatura: {temp_maxima}°C)")
```
